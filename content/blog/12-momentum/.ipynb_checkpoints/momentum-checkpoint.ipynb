```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('bmh') 
colors = ["#3d5a80","#98c1d9","#e0fbfc","#ee6c4d","#293241"]

def f(x, y):
    return 0.5 * x**2 + 10 * y**2

def grad(x, y):
    return np.array([x, 20 * y])

def gradient_descent(start, lr=0.09, beta=0.0, steps=20):
    pos = np.array(start, dtype=float)
    velocity = np.zeros(2)
    path = [pos.copy()]
    for _ in range(steps):
        velocity = beta * velocity - lr * grad(*pos)
        pos = pos + velocity
        path.append(pos.copy())
    return np.array(path)

def plot_descent(ax, path, title, path_color):
    X, Y = np.meshgrid(np.linspace(-10, 10, 100), np.linspace(-3.5, 3.5, 100))
    Z = f(X, Y)

    ax.plot_surface(X, Y, Z, cmap='Blues', alpha=0.55, linewidth=0, antialiased=True)
    ax.plot(path[:, 0], path[:, 1], f(path[:, 0], path[:, 1]) + 2,
            color=path_color, linewidth=2, marker='o', markersize=4)
    ax.set_title(title)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')
    ax.view_init(elev=35, azim=-70)
```

```python
path_plain = gradient_descent(start=(-8, 2.5), lr=0.09, beta=0.0)
path_momentum = gradient_descent(start=(-8, 2.5), lr=0.09, beta=0.2)

fig = plt.figure(figsize=(14, 6))
plot_descent(fig.add_subplot(1, 2, 1, projection='3d'), path_plain,
             'Gradient descent without momentum', colors[3])
plot_descent(fig.add_subplot(1, 2, 2, projection='3d'), path_momentum,
             'Gradient descent with momentum (β = 0.2)', colors[3])
plt.tight_layout()
plt.show()
```
