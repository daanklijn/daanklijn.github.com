```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('bmh') 
colors = ["#3d5a80","#98c1d9","#e0fbfc","#ee6c4d","#293241"]

X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# Add a row of 1s for bias.
X_with_bias = np.c_[X, np.ones(X.shape[0])]

y = np.array([[0], 
              [1], 
              [1], 
              [0]])

# Solve normal equations
xTx = np.dot(np.transpose(X_with_bias), X_with_bias)
xTy = np.dot(np.transpose(X_with_bias), y)
weights = np.dot(np.linalg.inv(xTx),xTy).round(2)

print("weights:", weights)


print("prediction:", np.dot(X_with_bias, weights))

```

```python
# NN with one hidden layer and two hidden units
# First hidden layer: h1= g(W_1*X + b_1)
# final layer: y = W_2*h1 + b_2

def relu(x):
    return np.maximum(0, x)

# Hidden layer 1 (2 inputs, 2 nodes)
W_h1 = np.random.randn(2, 2)
b_h1 = np.zeros(2)

# Output layer (2 inputs, 1 output)
W_o = np.random.randn(2,1)
b_o = np.zeros(1)

# Inference
h1 = relu(np.dot(X, W_h1) + b_h1)
y_hat = np.dot(h1, W_o) + b_o
y_hat
```
